```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    "AnnualIncome": [15, 18, 20, 22, 25,
                     70, 75, 78, 80, 85],
    "SpendingScore": [80, 85, 78, 90, 82,
                      20, 25, 15, 22, 18]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    df["AnnualIncome"],
    df["SpendingScore"]
)

plt.xlabel("Annual Income (in thousands)")
plt.ylabel("Spending Score")
plt.title("Customer Data Before Clustering")

plt.show()
```

```python
X = df[["AnnualIncome", "SpendingScore"]]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled.round(2))
```

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

kmeans.fit(X_scaled)

df["Cluster"] = kmeans.labels_

print(df)
```

```python
centroids_scaled = kmeans.cluster_centers_

centroids_original = scaler.inverse_transform(
    centroids_scaled
)

print("Cluster centroids in original units:")
print(centroids_original.round(2))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["AnnualIncome"],
    df["SpendingScore"],
    c=df["Cluster"],
    cmap="viridis",
    s=100
)

plt.scatter(
    centroids_original[:, 0],
    centroids_original[:, 1],
    color="red",
    marker="X",
    s=250,
    label="Centroids"
)

plt.xlabel("Annual Income (in thousands)")
plt.ylabel("Spending Score")
plt.title("K-Means Customer Segmentation (K=2)")
plt.legend()
plt.show()
```
