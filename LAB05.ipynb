```python
#Import Libraries
import numpy as np
import matplotlib.pyplot as plt

from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
```

```python
#Generate normal data
np.random.seed(42)

# Normal cluster 1
cluster1 = np.random.multivariate_normal(
    mean=[2, 2],
    cov=[[1, 0.5],
         [0.5, 1]],
    size=250
)

# Normal cluster 2
cluster2 = np.random.multivariate_normal(
    mean=[7, 6],
    cov=[[1, -0.3],
         [-0.3, 1]],
    size=250
)

# Combine normal data
X_normal = np.vstack([cluster1, cluster2])
```

```python
#Generate anomaly points
anomalies = np.array([
    [-2, 7],
    [1, 8],
    [10, 2],
    [11, 8],
    [5, 10],
    [9, 10],
    [-1, 4],
    [12, 5]
])

# Combine normal + anomaly data
X = np.vstack([X_normal, anomalies])
```

```python
#Fit GMM on normal data
gmm = GaussianMixture(
    n_components=2,
    covariance_type="full",
    random_state=42
)

gmm.fit(X_normal)
```

```python
#Calculate log-likelihood
scores = gmm.score_samples(X)

# Higher score = more likely to belong to normal distribution
# Lower score = more likely to be an anomaly
```

```python
#Set anomaly threshold
threshold = np.percentile(scores, 2)

print("Anomaly threshold:", threshold)
```

```python
#Detect anomalies
predicted_anomalies = scores < threshold

print("Number of detected anomalies:",
      predicted_anomalies.sum())
```

```python
#Visualize GMM anomaly detection
plt.figure(figsize=(10, 7))

# Normal points
plt.scatter(
    X[~predicted_anomalies, 0],
    X[~predicted_anomalies, 1],
    label="Normal",
    alpha=0.6
)

# Detected anomalies
plt.scatter(
    X[predicted_anomalies, 0],
    X[predicted_anomalies, 1],
    marker="X",
    s=120,
    label="Anomaly"
)

# Plot GMM means
plt.scatter(
    gmm.means_[:, 0],
    gmm.means_[:, 1],
    marker="*",
    s=250,
    label="GMM Centers"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("GMM-Based Anomaly Detection")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```
